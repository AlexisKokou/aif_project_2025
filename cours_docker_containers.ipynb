## 1. 🎯 Les Concepts de Base

### Qu'est-ce que Docker ?

Docker est une **plateforme de containerisation** qui permet de :
- **Empaqueter** une application avec toutes ses dépendances
- **Déployer** facilement sur n'importe quel environnement
- **Isoler** les applications les unes des autres
- **Standardiser** le processus de déploiement

### Pourquoi utiliser Docker ?

#### 🔥 **Problèmes sans Docker :**
- "Ça marche sur ma machine mais pas en production"
- Conflits de versions entre projets
- Installations complexes et longues
- Environnements de développement incohérents

#### ✅ **Avantages avec Docker :**
- **Portabilité** : Même comportement partout
- **Isolation** : Chaque application dans son propre environnement
- **Rapidité** : Démarrage instantané
- **Efficacité** : Partage des ressources système

## 2. 📦 Images vs Containers : La Différence Fondamentale

### 🖼️ **L'Image Docker**

Une image est comme une **"recette de cuisine"** ou un **"moule à gâteau"** :

- 📖 **Modèle/Template** : Instructions pour créer quelque chose
- 🔒 **Immuable** : Une fois créée, elle ne change plus
- 💾 **Stockée** : Sauvegardée sur le disque
- 🔄 **Réutilisable** : Peut créer plusieurs containers

```
Image = Blueprint (Plan architectural)
```

### 🏃‍♂️ **Le Container Docker**

Un container est comme **"le gâteau cuit"** ou **"la maison construite"** :

- ⚡ **Instance en cours d'exécution** d'une image
- 🔄 **Modifiable** : Peut être démarré, arrêté, modifié
- 🏠 **Isolé** : Possède son propre environnement
- 💨 **Temporaire** : Peut être supprimé facilement

```
Container = Instance vivante de l'image
```

### 🎭 Analogie Simple

| Concept | Image | Container |
|---------|-------|-----------|
| **Cuisine** | Recette de cuisine | Plat cuisiné |
| **Architecture** | Plans de maison | Maison habitée |
| **Musique** | Partition | Concert en cours |
| **Programmation** | Classe (Code) | Objet instancié |

### 🔄 Le Cycle de Vie Complet

```mermaid
graph TD
    A[Dockerfile] -->|docker build| B[Image]
    B -->|docker run| C[Container 1]
    B -->|docker run| D[Container 2]
    B -->|docker run| E[Container N]
    
    C -->|docker stop| F[Container arrêté]
    F -->|docker start| C
    F -->|docker rm| G[Supprimé]
```

**Exemple concret :**
1. **Dockerfile** = Instructions pour installer Python + notre app
2. **Image** = "Template" Python + notre app (stocké)
3. **Container** = Notre app qui tourne en live

## 3. 📝 Le Dockerfile : Comment Créer des Images

### Qu'est-ce qu'un Dockerfile ?

Un **Dockerfile** est un fichier texte contenant une série d'instructions pour construire une image Docker automatiquement.

### Instructions de Base

| Instruction | Description | Exemple |
|-------------|-------------|---------|
| `FROM` | Image de base | `FROM python:3.11-slim` |
| `WORKDIR` | Répertoire de travail | `WORKDIR /app` |
| `COPY` | Copier des fichiers | `COPY . /app` |
| `RUN` | Exécuter une commande | `RUN pip install -r requirements.txt` |
| `EXPOSE` | Exposer un port | `EXPOSE 8000` |
| `CMD` | Commande par défaut | `CMD ["python", "app.py"]` |

````python
# Exemple de Dockerfile pour une API Flask/FastAPI

FROM python:3.11-slim

# Installer les dépendances système si nécessaire
RUN apt-get update && apt-get install -y \
    libgl1-mesa-dev \
    libglib2.0-0 \
    && rm -rf /var/lib/apt/lists/*

# Définir le répertoire de travail
WORKDIR /app

# Copier et installer les dépendances Python
COPY requirements.txt .
RUN pip install --no-cache-dir --upgrade pip && \
    pip install --no-cache-dir -r requirements.txt

# Copier le code de l'application
COPY . .

# Exposer le port sur lequel l'app écoute
EXPOSE 5000

# Commande pour démarrer l'application
CMD ["python", "app.py"]
````

### 🎯 Explication Étape par Étape

1. **`FROM python:3.11-slim`** : On part d'une image Python légère
2. **`RUN apt-get update...`** : On installe les dépendances système nécessaires
3. **`WORKDIR /app`** : On se place dans le dossier `/app` du container
4. **`COPY requirements.txt .`** : On copie d'abord juste le fichier des dépendances
5. **`RUN pip install...`** : On installe les dépendances Python
6. **`COPY . .`** : On copie tout notre code
7. **`EXPOSE 5000`** : On indique que l'app utilise le port 5000
8. **`CMD ["python", "app.py"]`** : Commande pour démarrer l'app

### 💡 Pourquoi copier requirements.txt en premier ?

**Cache des layers** : Si le code change mais pas les dépendances, Docker peut réutiliser la couche avec `pip install` déjà exécutée !

## 4. ⚡ Commandes Essentielles Docker

### 🔧 Commandes pour les Images

````python
# 🔧 GESTION DES IMAGES

# Construire une image à partir d'un Dockerfile
docker build -t mon-app:latest .

# Construire avec un Dockerfile spécifique
docker build -f Dockerfile-api -t mon-api:latest .

# Lister toutes les images
docker images

# Supprimer une image
docker rmi mon-app:latest

# Supprimer toutes les images non utilisées
docker image prune

# Télécharger une image depuis Docker Hub
docker pull python:3.11

# Voir l'historique des couches d'une image
docker history mon-app:latest
````

````python
# 🏃‍♂️ GESTION DES CONTAINERS

# Créer et démarrer un container
docker run -d -p 8000:8000 --name mon-container mon-app:latest

# Démarrer un container en mode interactif
docker run -it --name debug-container python:3.11 bash

# Lister les containers actifs
docker ps

# Lister tous les containers (actifs et arrêtés)
docker ps -a

# Arrêter un container
docker stop mon-container

# Démarrer un container arrêté
docker start mon-container

# Redémarrer un container
docker restart mon-container

# Supprimer un container
docker rm mon-container

# Supprimer un container en cours d'exécution (forcé)
docker rm -f mon-container

# Voir les logs d'un container
docker logs mon-container

# Suivre les logs en temps réel
docker logs -f mon-container

# Exécuter une commande dans un container en cours d'exécution
docker exec -it mon-container bash
````

### 📋 Explication des Paramètres Importants

| Paramètre | Description | Exemple |
|-----------|-------------|---------|
| `-d` | Détaché (en arrière-plan) | `docker run -d` |
| `-it` | Interactif + terminal | `docker run -it` |
| `-p` | Mapping de ports | `-p 8000:5000` (hôte:container) |
| `--name` | Nom du container | `--name mon-app` |
| `-v` | Volume/montage | `-v /host:/container` |
| `--rm` | Supprime auto après arrêt | `docker run --rm` |
| `-e` | Variable d'environnement | `-e DB_HOST=localhost` |

## 5. 🎼 Docker Compose : Orchestrer Plusieurs Services

### Qu'est-ce que Docker Compose ?

Docker Compose permet de définir et gérer des **applications multi-containers** avec un seul fichier YAML.

### Avantages de Docker Compose

- 🎯 **Un seul fichier** pour définir tous les services
- 🚀 **Démarrage simple** avec `docker-compose up`
- 🔗 **Networking automatique** entre containers
- 📁 **Volumes partagés** facilement configurés
- 🔄 **Redémarrage automatique** en cas d'erreur

````python
# Exemple docker-compose.yml pour notre projet de films

services:
  # Service API (classification de films)
  api:
    build:
      context: .
      dockerfile: Dockerfile-api
    ports:
      - "5075:5075"
    volumes:
      - ./weights:/app/weights
    environment:
      - FLASK_ENV=production
    restart: unless-stopped
    
  # Service Interface Gradio
  gradio-app:
    build:
      context: .
      dockerfile: Dockerfile-gradio
    ports:
      - "7860:7860"
    environment:
      - API_URL=http://api:5075
    depends_on:
      - api
    restart: unless-stopped
    
  # Base de données (optionnel)
  database:
    image: postgres:15
    environment:
      POSTGRES_DB: movies_db
      POSTGRES_USER: admin
      POSTGRES_PASSWORD: password123
    volumes:
      - postgres_data:/var/lib/postgresql/data
    ports:
      - "5432:5432"

# Volumes nommés pour la persistance
volumes:
  postgres_data:
````

````python
# 🎼 COMMANDES DOCKER COMPOSE

# Démarrer tous les services
docker-compose up

# Démarrer en arrière-plan
docker-compose up -d

# Construire les images et démarrer
docker-compose up --build

# Arrêter tous les services
docker-compose down

# Arrêter et supprimer les volumes
docker-compose down -v

# Voir les logs de tous les services
docker-compose logs

# Voir les logs d'un service spécifique
docker-compose logs api

# Suivre les logs en temps réel
docker-compose logs -f

# Lister les services en cours
docker-compose ps

# Redémarrer un service spécifique
docker-compose restart api

# Exécuter une commande dans un service
docker-compose exec api bash
````

## 6. 🛡️ Bonnes Pratiques et Optimisations

### 🚀 Optimisation des Images

1. **Utilisez des images de base légères**
   ```dockerfile
   FROM python:3.11-slim  # ✅ Bien (~120MB)
   FROM python:3.11       # ❌ Lourd (~900MB)
   ```

2. **Multi-stage builds** pour réduire la taille finale
   ```dockerfile
   # Stage de build
   FROM python:3.11 as builder
   WORKDIR /app
   COPY requirements.txt .
   RUN pip install --user -r requirements.txt
   
   # Stage final
   FROM python:3.11-slim
   WORKDIR /app
   COPY --from=builder /root/.local /root/.local
   COPY . .
   ```

3. **Ordre optimal des instructions** (du moins changeant au plus changeant)
   ```dockerfile
   FROM python:3.11-slim          # 1. Image de base (jamais)
   RUN apt-get update...          # 2. Dépendances système (rarement)
   COPY requirements.txt .        # 3. Fichier de deps (parfois)
   RUN pip install...             # 4. Installation deps (parfois)
   COPY . .                       # 5. Code source (souvent)
   ```

### 🔒 Sécurité

1. **Ne jamais inclure de secrets dans l'image**
   ```dockerfile
   # ❌ Mauvais
   ENV API_KEY=secret123
   
   # ✅ Bon - utiliser des variables d'env à l'exécution
   docker run -e API_KEY=secret123 mon-app
   ```

2. **Utilisez un utilisateur non-root**
   ```dockerfile
   RUN addgroup --system app && adduser --system --group app
   USER app
   ```

3. **Fichier .dockerignore**
   ```
   .git
   node_modules
   __pycache__
   *.pyc
   .env
   .DS_Store
   ```

### 🧹 Nettoyage et Maintenance

```bash
# Nettoyer les containers arrêtés
docker container prune

# Nettoyer les images non utilisées
docker image prune

# Nettoyer les volumes non utilisés
docker volume prune

# Nettoyer tout le système Docker
docker system prune -a
```
